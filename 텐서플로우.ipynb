```python
# 다차원 배열 - ndarray - numpy
# numpy Array(다차원 어레이) + 기능 = tensor 
```

```python
import tensorflow as tf

a = tf.constant(7.0, name = 'data1')
b = tf.constant(3.0, name = 'data2')
c = tf.constant(2.0, name = 'data3')

v = a * b / c

with tf.Session() as sess:
    sess.run(tf.global_variables_initializer())
    print(sess.run(v))
```

```python
import tensorflow.compat.v1 as tf # 1.x버전을 임포트
tf.disable_v2_behavior(); # 2.x 버전을 비활성화

a = tf.constant(7.0, name = 'data1')
b = tf.constant(3.0, name = 'data2')
c = tf.constant(2.0, name = 'data3')

v = a * b / c

with tf.Session() as sess:
    sess.run(tf.global_variables_initializer())
    print(sess.run(v))
```

```python
# 즉시 실행이라는 개념을 하지 않고 지연 실행을 한다. v1.x에서는 
# 지연 실행보다 즉시 실행을 해주는 파이토치가 인기가 훨씬 많아진다.
```

```python
import tensorflow as tf

# 상수 정의
a = tf.constant(5)
b = tf.constant(2)
c = tf.constant(3)

# 연산
d = tf.multiply(a,b)
e = tf.add(c,b)
f = tf.subtract(d,e)

# 값출력
print("a =", a.numpy())
# .numpy()를 쓰면 실세 숫자 값 출력
print("b =", b.numpy())
print("c =", c.numpy())
print("d =", d.numpy())
print("e =", e.numpy())
print("f =", f.numpy())
```

```python
# 입력과 가중치 정의(tf.Variable로 지정, 값 변경 가능)
I = tf.Variable([[1,2,3,4], [5,6,7,8]], dtype=tf.float32)
W = tf.Variable([[1,1], [2,2], [3,3], [4,4]], dtype=tf.float32)

# 행렬곱 계산
m_cal = tf.matmul(I,W)

# 결과 출력
print(m_cal.numpy())
```

## 1. 기본 실행 환경 확인

현재 주피터가 어떤 파이썬 환경에서 실행 중인지 확인합니다.

```python
# 기본 라이브러리를 불러옵니다.
from pathlib import Path
import sys
import platform

# 현재 파이썬 실행 파일과 버전을 확인합니다.
print("Python 실행 경로:", sys.executable)
print("Python 버전:", sys.version)
print("운영체제:", platform.platform())
```

## 2. 프로젝트 폴더 확인

`final_project`를 기준 폴더로 잡고, 이후 모든 데이터 경로를 이 폴더 아래에서 관리합니다.

```python
# 현재 노트북 실행 위치를 확인합니다.
current_dir = Path.cwd()

# 주피터를 루트 폴더에서 열었는지, final_project 폴더에서 열었는지 모두 대응합니다.
if current_dir.name == "final_project":
    project_dir = current_dir
elif (current_dir / "final_project").exists():
    project_dir = current_dir / "final_project"
else:
    project_dir = current_dir

print("현재 실행 위치:", current_dir)
print("프로젝트 폴더:", project_dir)
print("프로젝트 폴더 존재 여부:", project_dir.exists())
```

## 3. 작업 폴더 생성 및 확인

원천 데이터, 가공 데이터, 모델, 결과물을 저장할 폴더를 확인하고 없으면 생성합니다.

```python
# 프로젝트에서 사용할 기본 폴더 목록입니다.
data_raw_dir = project_dir / "data" / "raw"
data_processed_dir = project_dir / "data" / "processed"
models_dir = project_dir / "models"
outputs_dir = project_dir / "outputs"

project_dirs = [
    data_raw_dir,
    data_processed_dir,
    models_dir,
    outputs_dir,
]

# 폴더가 없으면 생성합니다.
for folder in project_dirs:
    folder.mkdir(parents=True, exist_ok=True)

# 생성 또는 확인된 폴더를 출력합니다.
for folder in project_dirs:
    print(folder, "->", "존재함" if folder.exists() else "없음")
```

## 5. 생성된 `transactions.csv` 확인

나중에 데이터 생성 노트북을 실행한 뒤, `data/processed/transactions.csv` 파일이 제대로 만들어졌는지 확인합니다.

```python
# 최종 가공 데이터 파일 경로를 지정합니다.
# 앞 셀을 실행하지 않고 이 셀만 실행해도 동작하도록 기본 경로를 보강합니다.
from pathlib import Path

if "project_dir" not in globals():
    current_dir = Path.cwd()
    if current_dir.name == "final_project":
        project_dir = current_dir
    elif (current_dir / "final_project").exists():
        project_dir = current_dir / "final_project"
    else:
        project_dir = Path("/Users/gwongwangjae/goorm-ai-language-course/final_project")

if "data_processed_dir" not in globals():
    data_processed_dir = project_dir / "data" / "processed"

transactions_path = data_processed_dir / "transactions.csv"

# transactions.csv 파일이 있는지 확인합니다.
print("transactions.csv 경로:", transactions_path)
print("파일 존재 여부:", transactions_path.exists())

if transactions_path.exists():
    # 파일이 있으면 크기를 MB 단위로 확인합니다.
    file_size_mb = transactions_path.stat().st_size / (1024 * 1024)
    print(f"파일 크기: {file_size_mb:.2f} MB")
else:
    print("아직 transactions.csv가 없습니다. 데이터 생성 노트북을 실행한 뒤 다시 확인하세요.")
```

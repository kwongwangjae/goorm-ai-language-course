```python
import numpy as np

# 데이터 정의 (입력 데이터와 출력 데이터)
data_x = np.array([10, 20, 30, 40])
data_y = np.array([21, 49, 68, 80])

# 평균값 계산
mean_x = np.mean(data_x)
mean_y = np.mean(data_y)

# 기울기 계산 
numerator = np.sum((data_x - mean_x) * (data_y - mean_y))
denominator = np.sum((data_x - mean_x) ** 2)
slope = numerator / denominator

# 절편 계산
intercept = mean_y - slope * mean_x

# 결과 출력
print("기울기:", slope)
print("절편:", intercept)
```

```python
# 아무 기울기 선을 막 긋다가 그 선이 데이터를 가장 잘 표현할 떄 멈추라
# 그럼 이때 "잘 표현한다는 것이 무엇인가?"
```

```python
# 데이터를 가장 잘 표현한다. -> 예측값과 실체값의 오차를 줄인다
# 오차(실체값과 예측값의 차이)를 줄이기 위해 파라미터 값(y = ax + b)을 변경하는 것 -> 이것이 딥러닝 모델링이다.
# 오차가 클수록 a와 b를 찾아가는데에서 편하다. -> 오차를 크게하는 법(제곱 해보기) loss = 'mse' 
```

```python
# numpy는 백터 연산이 일어난다.
# 오차범위가 4 정도면 나쁘지 않은 코드다.
import numpy as np

# 실제값과 예측값
y_true = [21,49,68,80]
y_pred = [25.1, 44.7, 64.3, 83.9]

# RMSE 계산 함수
# ** 는 거듭제곱 연산자
# 2 ** 3 -> 2의 3제곱 = 8
def rmse(y_true, y_pred):
    # np.mean은 NumPy에서 평균값을 구하는 함수
    mse = np.mean((np.array(y_true) - np.array(y_pred)) ** 2)
    return np.sqrt(mse)

# 결과 출력
print("RMSE:", rmse(y_true, y_pred))
```

```python
# 파이썬 코드로 풀면??

# 실제값과 예측값
y_true = [21, 49, 68, 80]
y_pred = [25.1, 44.7, 64.3, 83.9]

# RMSE 계산 함수
def rmse(y_true, y_pred):
    total = 0

    for true, pred in zip(y_true, y_pred):
        error = true - pred
        squared_error = error ** 2
        total += squared_error

    mse = total / len(y_true)
    return mse ** 0.5

# 결과 출력
print("RMSE:", rmse(y_true, y_pred))
```

```python
# 1. 공부시간 (x축)과 점수 (y축) 간의 그래프(scatter) 작성

import numpy as np
import matplotlib.pyplot as plt

# 공부 시간 x와 성적 y의 리스트를 만들기
data = [[2, 81], [4, 93], [6,91], [8,97]]
x = [i[0] for i in data]
y = [i[1] for i in data]

# x,y data 확인
print(x,y)

# 그래프로 나타내기
plt.figure(figsize=(8,5))
plt.scatter(x,y)
plt.show()
```

```python
x_data = np.array(x)
y_data = np.array(y)

# y = ax + b
a=0
b=0

# 0.05를 했을 때 발산을 했다. -> 0.03으로 줄이기
# 핵심은 lr(런닝 레이트) -> 하이퍼 파라미터(우리가 계속 값을 바꿔가면서 추적해가는 값)
lr = 0.03

# 2000번을 하강함 -> 몇번을 해야하냐? -> keras에서 할 수 있다.
epochs = 2001

for i in range(epochs):
    # 오차함수를 a로 미분한 값
    a_diff = 2/len(x_data) * sum((a * x_data + b - y_data)* x_data)
    # 오차함수를 b로 미분한 값
    b_diff = 2/len(x_data) * sum(a * x_data + b - y_data)
    a = a - lr * a_diff # 기울기 업데이트
    b = b - lr * b_diff # 절편 업데이트
    if i % 100 == 0:
        print("epoch=%.f, 기울기=%.04f, 절편=%.04f" % (i, a, b))
```

```python
y_pred = a * x_data + b
plt.scatter(x,y)
plt.plot(x_data,y_pred)
plt.show()
```

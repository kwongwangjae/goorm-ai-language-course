## 1. `print()` 함수

파이썬의 기준 문법(reference)은 Python Software Foundation이 관리하는 공식 사이트 
**Python**의 문서인 **Python Documentation**에 정리된 내용이 기준

https://docs.python.org/ko/3.14/library/functions.html#print

`print()` 함수는 Python에서 화면에 데이터를 출력할 때 사용됩니다.  
가장 많이 쓰이는 내장 함수(Built-in Function) 중 하나입니다.

**기본 형태:**
```python
print(value1, value2, ..., sep=' ', end='\n', file=sys.stdout)
```

| 매개변수 | 설명 | 기본값 |
|---------|------|-------|
| `value` | 출력할 값 (여러 개 가능) | - |
| `sep` | 값과 값 사이에 넣을 문자 | 공백(' ') |
| `end` | 출력이 끝난 후 붙는 문자 | 줄바꿈('\\n') |
| `file` | 출력 대상 (파일 등) | 화면(stdout) |

````python
print("Hello, World!")
print('Hello, World!')
````

### 1-1. 다양한 자료형 출력하기

`print()` 함수는 숫자, 문자열, 리스트 등 Python의 모든 자료형을 출력할 수 있습니다.  
내부적으로 각 객체의 `__str__()` 메서드를 호출하여 문자열로 변환 후 출력합니다.

````python
print("42"+"35")
print(42+35)
````

````python
# 정수(int) 출력
print(42)

# 실수(float) 출력
print(3.14)

# 문자열(str) 출력
print("Python is fun!")

# 불리언(bool) 출력
print(True)
print(False)

# 리스트(list) 출력
print([1, 2, 3, 4, 5])

# 딕셔너리(dict) 출력
print({"name": "Alice", "age": 25})

# 여러 값을 한 번에 출력 (기본: 공백으로 구분)
print("나이:", 25, "점수:", 99.5)
````

### 1-2. `sep`과 `end` 매개변수

- **`sep`**: 여러 값을 출력할 때 값 사이에 삽입되는 구분자입니다. 기본값은 공백(' ')입니다.
- **`end`**: 출력이 끝난 후 마지막에 붙는 문자입니다. 기본값은 줄바꿈('\\n')입니다.

> 💡 **핵심 개념**: `print()`는 기본적으로 출력 후 자동으로 줄을 바꿉니다.  
> `end=""`로 설정하면 줄바꿈 없이 이어서 출력할 수 있습니다.

````python
# sep 기본값(공백)
print("사과", "바나나", "체리")          # 사과 바나나 체리

# sep를 쉼표+공백으로 변경
print("사과", "바나나", "체리", sep=", ")  # 사과,바나나,체리

# sep를 화살표로 변경
print("1단계", "2단계", "3단계", sep=" → ")  # 1단계 → 2단계 → 3단계

# sep를 줄바꿈으로 설정 (세로로 출력)
print("월", "화", "수", "목", "금", sep="\n")

print("-" * 30)

# end 기본값(줄바꿈)
print("첫 번째 줄")   # 출력 후 자동으로 줄바꿈
print("두 번째 줄")

# end를 빈 문자열로 설정 (줄바꿈 없음)
print("Hello", end="")
print("World!")         # HelloWorld!

# end를 공백으로 설정
print("A", end="")
print("B", end="")
print("C")              # A B C
````

### 변수의 기본 개념  
1. 변수란?  
**변수**(variable)는 **데이터를 저장하는 이름이 붙은 공간**  
쉽게 말하면 **값을 보관하는 상자**  
예

설명

| ------- | --------- |

| `"홍길동"` | 저장되는 데이터  |  
  
name → "홍길동" 으로 이해하면 됨

의미

변수 `age`에 **20이라는 값이 저장**

### 1-3. 문자열 포매팅 (f-string, format, %)

변수 값을 문자열에 삽입하는 방법은 세 가지가 있습니다.

| 방식 | 형태 | 특징 |
|------|------|------|
| `%` 연산자 | `"%s" % 변수` | 구버전 방식, C언어 스타일 |
| `.format()` | `"{}".format(변수)` | Python 2.6부터 지원 |
| f-string | `f"{변수}"` | Python 3.6부터 지원, 가장 권장 |

> 💡 **현재 가장 많이 사용되고 권장되는 방식은 f-string입니다.**

````python
name = "홍길동"
age = 25
score = 98.5
````

````python
# 방법 1: % 연산자 (구버전 방식)
print("이름: %s, 나이: %d" % (name, age))

# 방법 2: .format() 메서드
print("이름: {}, 나이: {}".format(name, age))
print("이름: {0}, 나이: {1}, 점수: {2:.1f}".format(name, age, score))  # 인덱스 및 소수점 지정
````

````python
# 방법 3: f-string (Python 3.6+, 가장 권장)
print(f"이름: {name}, 나이: {age}")
print(f"점수: {score:.2f}")         # 소수점 2자리
print(f"10 + 20 = {10 + 20}")      # 표현식(수식)도 사용 가능
print(f"이름 길이: {len(name)}자")   # 함수 호출도 가능
````

### 2-1. 숫자 입력 받기 - 형 변환(Type Casting)

Python의 주요 형 변환 함수:

| 함수 | 변환 대상 | 예시 |
|------|----------|------|
| `int()` | 정수로 변환 | `int("10")` → `10` |
| `float()` | 실수로 변환 | `float("3.14")` → `3.14` |
| `str()` | 문자열로 변환 | `str(42)` → `"42"` |
| `bool()` | 불리언으로 변환 | `bool(0)` → `False` |

````python
# 위 예제의 올바른 정답 코드 확인
data1 = int(input("첫번째 숫자 입력 :"))
data2 = int(input("두번째 숫자 입력 :"))
print(f"{data1}과 {data2}을 입력하였습니다. 두수의 합은 {data1+data2}입니다.")
````

````python
# input()은 항상 문자열 반환 → 타입 확인
raw = input("숫자를 입력하세요: ")
print(f"입력값: {raw}, 타입: {type(raw)}")  # <class 'str'>

# int()로 변환
num = int(raw)
print(f"변환 후: {num}, 타입: {type(num)}")  # <class 'int'>

# 한 줄로 작성하는 방법
age = int(input("나이를 입력하세요: "))
print(f"당신은 {age}살이고, 10년 후엔 {age + 10}살입니다.")

# 실수 입력
height = float(input("키를 입력하세요(cm): "))
print(f"키: {height}cm, m로 환산: {height/100:.2f}m")
````

### 2-2. 여러 값 입력 받기 - `split()`과 `map()`

- **`split(구분자)`**: 문자열을 구분자 기준으로 나눠 리스트로 반환합니다. 기본 구분자는 공백입니다.
- **`map(함수, 이터러블)`**: 리스트의 각 요소에 함수를 적용합니다.

```
"1 2 3".split()  →  ['1', '2', '3']          (문자열 리스트)
map(int, ['1','2','3'])  →  [1, 2, 3]         (정수 리스트)
```

````python
data1, data2, data3 = input("세 개의 숫자를 공백으로 구분해 입력: ").split()
print(int(data1)+int(data2)+int(data3))
````

````python
data1, data2, data3 = map(float, input("세 개의 숫자를 공백으로 구분해 입력: ").split())
#data1, data2, data3 = map(int, input("세 개의 숫자를 공백으로 구분해 입력: ").split())
print(data1+data2+data3)
````

````python
# split() 기본 사용 - 공백 기준 분리
data = input("세 개의 단어를 공백으로 구분해 입력: ").split()
print("리스트:", data)               # ['단어1', '단어2', '단어3']
print("첫 번째 단어:", data[0])

# split() - 쉼표 기준 분리
fruits = input("과일을 쉼표로 구분해 입력(예: 사과,바나나,딸기): ").split(",")
print("과일 목록:", fruits)

# map() + split() - 정수 여러 개 입력
a, b = map(int, input("두 정수를 공백으로 입력: ").split())
print(f"{a} + {b} = {a + b}")
print(f"{a} × {b} = {a * b}")

# 리스트로 받기
numbers = list(map(int, input("정수 여러 개 입력: ").split()))
print("입력한 숫자:", numbers)
print("합계:", sum(numbers))
print("최댓값:", max(numbers))
print("최솟값:", min(numbers))
````

## 3. 변수(Variable)

변수는 **데이터를 저장하는 이름이 붙은 메모리 공간**입니다.

Python 변수의 특징:
- **동적 타입(Dynamic Typing)**: 선언 시 타입을 지정하지 않습니다.
- **타입은 값에 따라 자동 결정**됩니다.
- 같은 변수에 다른 타입의 값도 재할당할 수 있습니다.

**변수 이름 규칙:**
- 영문자, 숫자, 언더스코어(`_`)만 사용 가능
- 숫자로 시작할 수 없음
- 파이썬 예약어(if, for, while 등) 사용 불가
- 대소문자 구분 (`name`과 `Name`은 다른 변수)

### 3-1. 변수 선언 및 자료형

````python
# 정수(int): 소수점 없는 숫자
age = 25
score = -10
print(f"age = {age}, type = {type(age)}")

# 실수(float): 소수점 있는 숫자
pi = 3.14159
temperature = -2.5
print(f"pi = {pi}, type = {type(pi)}")

# 문자열(str): 따옴표로 감싼 텍스트
name = "홍길동"
message = '안녕하세요!'
multiline = """여러 줄
문자열입니다."""
print(f"name = {name}, type = {type(name)}")

# 불리언(bool): True 또는 False
is_student = True
is_adult = False
print(f"is_student = {is_student}, type = {type(is_student)}")

# None: 값이 없음을 나타냄
result = None
print(f"result = {result}, type = {type(result)}")
````

### 3-2. 변수 값 변경과 동적 타입

````python
# 변수 값 변경
x = 10
print(f"처음: x = {x}, type = {type(x)}")

x = 20
print(f"변경 후: x = {x}")

# Python은 동적 타입 - 같은 변수에 다른 타입 할당 가능
x = "안녕하세요"
print(f"문자열로: x = {x}, type = {type(x)}")

x = 3.14
print(f"실수로: x = {x}, type = {type(x)}")

print()

# 복합 대입 연산자 (축약형)
count = 0
count += 1    # count = count + 1
count += 1
print(f"count = {count}")   # 2

count -= 1    # count = count - 1
print(f"count = {count}")   # 1

count *= 5    # count = count * 5
print(f"count = {count}")   # 5

count //= 2   # count = count // 2 (정수 나눗셈)
print(f"count = {count}")   # 2

count = 10
count %= 3   # count = count % 3 (정수 나머지 연산)
print(f"count = {count}")   # 2
````

### 3-3. 다중 할당과 변수 교환

````python
# 여러 변수 동시 할당
x, y, z = 1, 2, 3
print(f"x={x}, y={y}, z={z}")

# 같은 값을 여러 변수에 할당
a = b = c = 0
print(f"a={a}, b={b}, c={c}")

print()

# 변수 값 교환 (Python만의 간결한 문법)
a = 5
b = 10
print(f"교환 전: a={a}, b={b}")

# 다른 언어에서는 임시 변수가 필요:
# temp = a
# a = b
# b = temp

# Python에서는 한 줄로 해결!
a, b = b, a
print(f"교환 후: a={a}, b={b}")

print()

# 연산 결과를 변수에 저장
width = 10
height = 5
area = width * height
perimeter = 2 * (width + height)
print(f"직사각형 - 너비:{width}, 높이:{height}")
print(f"넓이: {area}, 둘레: {perimeter}")
````

### 3-4. 산술 연산자

| 연산자 | 의미 | 예시 | 결과 |
|--------|------|------|------|
| `+` | 덧셈 | `7 + 3` | `10` |
| `-` | 뺄셈 | `7 - 3` | `4` |
| `*` | 곱셈 | `7 * 3` | `21` |
| `/` | 나눗셈 (실수) | `7 / 2` | `3.5` |
| `//` | 정수 나눗셈 (몫) | `7 // 2` | `3` |
| `%` | 나머지 | `7 % 2` | `1` |
| `**` | 거듭제곱 | `2 ** 10` | `1024` |

````python
a, b = 17, 5

print(f"{a} + {b}  = {a + b}")
print(f"{a} - {b}  = {a - b}")
print(f"{a} * {b}  = {a * b}")
print(f"{a} / {b}  = {a / b}")   # 실수 나눗셈
print(f"{a} // {b} = {a // b}")  # 몫 (정수 나눗셈)
print(f"{a} % {b}  = {a % b}")   # 나머지
print(f"2 ** 10 = {2 ** 10}")     # 거듭제곱

print()

# 활용 예시: 짝수/홀수 판별
n = 17
print(f"{n}은(는) {'짝수' if n % 2 == 0 else '홀수'}입니다.")

# 활용 예시: 시간 변환
total_seconds = 3725
hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60
print(f"{total_seconds}초 = {hours}시간 {minutes}분 {seconds}초")
````

사용자로부터 '총 주행 거리(km)'와 '사용한 연료량(L)'을 실수(float)로 입력받아,
차량의 연비(km/L)를 소수점 첫째 자리까지 f-string으로 출력하는 코드를 작성하세요.

````python
# 정답 예시
distance = float(input("총 주행 거리(km)를 입력하세요: "))
fuel = float(input("사용한 연료량(L)을 입력하세요: "))

efficiency = distance / fuel
print(f"차량의 연비는 {efficiency:.1f} km/L 입니다.")
````

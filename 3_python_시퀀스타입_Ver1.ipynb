# Python 시퀀스 타입(Sequence Types)

이 노트북에서는 Python의 시퀀스 자료형(리스트, 튜플, 문자열)과 딕셔너리에 대해 학습합니다.

**시퀀스(Sequence)** 란 데이터가 **순서(index)를 가지고 나열**된 자료형입니다.

| 자료형 | 특징 | 변경 가능 | 예시 |
|--------|------|:---------:|------|
| 리스트(list) | 순서 있음, 중복 허용 | ✅ 가능 | `[1, 2, 3]` |
| 튜플(tuple) | 순서 있음, 중복 허용 | ❌ 불가 | `(1, 2, 3)` |
| 문자열(str) | 순서 있음, 문자들의 나열 | ❌ 불가 | `"Hello"` |
| 딕셔너리(dict) | 키-값 쌍, 순서 있음(3.7+) | ✅ 가능 | `{"a": 1}` 시퀀스 아님|

## 1. 인덱싱(Indexing)과 슬라이싱(Slicing)

모든 시퀀스 자료형에 공통으로 사용되는 핵심 개념입니다.

### 인덱스(Index)
- 시퀀스의 각 요소에 접근하는 **위치 번호**입니다.
- **앞에서부터**: 0, 1, 2, 3, ...
- **뒤에서부터**: -1, -2, -3, ...

```
문자열:  H  e  l  l  o
양수:    0  1  2  3  4
음수:   -5 -4 -3 -2 -1
```

### 슬라이싱(Slicing)
```python
시퀀스[시작:끝:간격]
```
- `시작`: 포함 (생략 시 처음부터)
- `끝`: 미포함 (생략 시 끝까지)
- `간격`: 몇 칸씩 건너뛸지 (생략 시 1)

`````python
data = [10, 20, 30, 40, 50, 60, 70]
#        0   1   2   3   4   5   6
#       -7  -6  -5  -4  -3  -2  -1

# 인덱싱
print(data[0])    # 10  (첫 번째)
print(data[3])    # 40
print(data[-1])   # 70  (마지막)
print(data[-2])   # 60  (뒤에서 두 번째)

print()

# 슬라이싱
print(data[1:4])   # [20, 30, 40]  (인덱스 1,2,3)
print(data[:3])    # [10, 20, 30]  (처음~인덱스2)
print(data[4:])    # [50, 60, 70]  (인덱스4~끝)
print(data[:])     # [10, 20, 30, 40, 50, 60, 70]  (전체)
print(data[::2])   # [10, 30, 50, 70]  (2칸씩)
print(data[::-1])  # [70, 60, 50, 40, 30, 20, 10]  (역순)
print(data[1:-1])  # [20, 30, 40, 50, 60]  (첫, 마지막 제외)
`````

`````python
#range 간단 정리
my_l = list(range(10))
my_l1 = list(range(3,10))
my_l2 = list(range(3,10,2))
print(my_l)
print(my_l1)
print(my_l2)
`````

## 2. 리스트(List)

리스트는 **순서가 있고 변경 가능(mutable)** 한 자료형입니다.  
다양한 타입의 값을 함께 저장할 수 있습니다.

### 2-1. 리스트 생성

`````python
# 리스트 생성 방법들
empty = []                          # 빈 리스트
numbers = [1, 2, 3, 4, 5]          # 정수 리스트
fruits = ["apple", "banana", "cherry"]  # 문자열 리스트
mixed = [1, "hello", 3.14, True, None]  # 혼합 타입
nested = [[1, 2], [3, 4], [5, 6]]  # 중첩 리스트 (2차원)

print(numbers)
print(fruits)
print(mixed)
print(nested[1])      # [3, 4]
print(nested[1][0])   # 3 (2차원 접근)

# list() 함수로 생성
chars = list("Hello")         # ['H', 'e', 'l', 'l', 'o']
nums = list(range(1, 6))      # [1, 2, 3, 4, 5]
print(chars)
print(nums)
`````

### 2-2. 리스트 주요 메서드

| 메서드 | 설명 | 반환값 |
|--------|------|--------|
| `append(x)` | 끝에 요소 추가 | None |
| `insert(i, x)` | i번 위치에 요소 삽입 | None |
| `extend(list)` | 리스트 뒤에 다른 리스트 붙이기 | None |
| `remove(x)` | 값 x인 첫 번째 요소 삭제 | None |
| `pop(i)` | i번 위치 요소 삭제 및 반환 | 삭제된 값 |
| `index(x)` | x의 인덱스 반환 | 정수 |
| `count(x)` | x의 개수 반환 | 정수 |
| `sort()` | 정렬 (원본 변경) | None |
| `sorted()` | 정렬된 새 리스트 반환 | 리스트 |
| `reverse()` | 역순 정렬 | None |
| `len()` | 요소 개수 | 정수 |

`````python
fruits = ["apple", "banana", "cherry"]

# 추가
fruits.append("orange")         # 끝에 추가
fruits.insert(1, "grape")       # 1번 위치에 삽입
fruits.extend(["kiwi", "mango"])  # 리스트 연결
print("추가 후:", fruits)

# 삭제
fruits.remove("banana")         # 값으로 삭제
popped = fruits.pop()           # 마지막 요소 삭제 및 반환
print(f"삭제 후: {fruits}, pop됨: {popped}")

# 검색
idx = fruits.index("orange")
cnt = fruits.count("apple")
print(f"orange 위치: {idx}, apple 개수: {cnt}")

# 정렬
nums = [5, 2, 8, 1, 9, 3]
nums.sort()              # 오름차순 (원본 변경)
print("오름차순:", nums)
nums.sort(reverse=True)  # 내림차순 (원본 변경)
print("내림차순:", nums)

original = [5, 2, 8, 1]
sorted_new = sorted(original)  # 새 리스트 반환 (원본 유지)
print(f"원본: {original}, 정렬된 새 리스트: {sorted_new}")
`````

## 3. 튜플(Tuple)

튜플은 **순서가 있고 변경 불가능(immutable)** 한 자료형입니다.  
한 번 생성 후 수정이 불가하므로 **변경되면 안 되는 데이터**를 저장할 때 사용합니다.

`````python
# 튜플 생성
empty_t = ()                    # 빈 튜플
single = (42,)                  # 요소 1개 (쉼표 필수!)
point = (10, 20)                # 좌표
rgb = (255, 128, 0)             # RGB 색상
info = ("홍길동", 25, "서울")    # 혼합 타입

print(f"좌표: x={point[0]}, y={point[1]}")
print(f"RGB: R={rgb[0]}, G={rgb[1]}, B={rgb[2]}")

# 튜플 언패킹 (다중 할당)
name, age, city = info
print(f"이름: {name}, 나이: {age}, 도시: {city}")

# 변경 시 오류 발생!
try:
    point[0] = 100   # 오류!
except TypeError as e:
    print(f"오류: {e}")

# 튜플 ↔ 리스트 변환
t = (1, 2, 3)
l = list(t)      # 튜플 → 리스트 (변경 가능)
l.append(4)
t2 = tuple(l)    # 리스트 → 튜플
print(f"리스트: {l}, 튜플: {t2}")

# 튜플은 리스트보다 메모리 효율적이고 속도가 빠름
import sys
print(f"리스트 크기: {sys.getsizeof([1,2,3])} bytes")
print(f"튜플 크기: {sys.getsizeof((1,2,3))} bytes")
`````

## 4. 문자열(String)

문자열은 **문자들의 시퀀스**입니다. 따옴표로 감싸서 표현합니다.

### 4-1. 문자열 생성과 기본 연산

`````python
# 문자열 생성
s1 = "큰따옴표 문자열"
s2 = '작은따옴표 문자열'
s3 = """여러 줄
문자열입니다."""

# 문자열 연산
a = "Hello"
b = " World"

print(a + b)        # 연결: Hello World
print(a * 3)        # 반복: HelloHelloHello
print(len(a))       # 길이: 5
print("H" in a)     # 포함 여부: True

# 인덱싱과 슬라이싱 (리스트와 동일)
s = "Python"
print(s[0])         # P
print(s[-1])        # n
print(s[2:5])       # tho
print(s[::-1])      # nohtyP (역순)
`````

### 4-2. 문자열 주요 메서드

| 메서드 | 설명 | 예시 |
|--------|------|------|
| `upper()` | 대문자 변환 | `"hello".upper()` → `"HELLO"` |
| `lower()` | 소문자 변환 | `"HELLO".lower()` → `"hello"` |
| `strip()` | 앞뒤 공백 제거 | `" hi ".strip()` → `"hi"` |
| `split(sep)` | 구분자로 분리 | `"a,b,c".split(",")` → `['a','b','c']` |
| `join(list)` | 리스트를 문자열로 합치기 | `",".join(['a','b'])` → `"a,b"` |
| `replace(old, new)` | 문자열 치환 | `"aaa".replace("a","b")` → `"bbb"` |
| `find(sub)` | 위치 찾기 (없으면 -1) | `"hello".find("ll")` → `2` |
| `count(sub)` | 부분 문자열 개수 | `"hello".count("l")` → `2` |
| `startswith(s)` | 시작 문자열 확인 | `"hello".startswith("he")` → `True` |
| `endswith(s)` | 끝 문자열 확인 | `"hello".endswith("lo")` → `True` |

`````python
text = "  Hello, Python World!  "

# 공백 처리
print(text.strip())         # 앞뒤 공백 제거
print(text.lstrip())        # 왼쪽 공백만 제거
print(text.rstrip())        # 오른쪽 공백만 제거

# 대소문자
s = "hello python"
print(s.upper())            # HELLO PYTHON
print(s.capitalize())       # Hello python (첫 글자만 대문자)
print(s.title())            # Hello Python (각 단어 첫 글자)

# 분리와 합치기
csv = "사과,바나나,딸기,포도"
fruits = csv.split(",")
print(fruits)               # ['사과', '바나나', '딸기', '포도']
print(" / ".join(fruits))   # 사과 / 바나나 / 딸기 / 포도

# 검색과 치환
sentence = "I love Python. Python is great."
print(sentence.find("Python"))           # 7 (첫 번째 위치)
print(sentence.count("Python"))          # 2
print(sentence.replace("Python", "Java"))  # 모두 치환
print(sentence.startswith("I"))          # True
print(sentence.endswith("."))            # True

# 타입 확인
print("12345".isdigit())    # True: 모두 숫자
print("hello".isalpha())    # True: 모두 알파벳
print("hello123".isalnum()) # True: 알파벳+숫자
`````

## 5. 딕셔너리(Dictionary)

딕셔너리는 **키(key)-값(value) 쌍**으로 데이터를 저장하는 자료형입니다.  
키를 통해 빠르게 값을 검색할 수 있습니다.

- 키는 중복될 수 없으며, 변경 불가능한 타입(str, int, tuple 등)만 사용 가능
- 값은 어떤 타입이든 가능

`````python
# 딕셔너리 생성
#student = {"name": "홍길동","age": 20,"scores": [90, 85, 92],"is_active": True}
student = {
    "name": "홍길동",
    "age": 20,
    "scores": [90, 85, 92],
    "is_active": True
}
# 값 접근
print(student["name"])          # 홍길동
print(student.get("age1"))       # 20 (get: 키 없어도 오류 안 남)
print(student.get("age1","30"))       # 20 (get: 키 없어도 오류 안 남)
`````

`````python
# 딕셔너리 생성
#student = {"name": "홍길동","age": 20,"scores": [90, 85, 92],"is_active": True}
student = {
    "name": "홍길동",
    "age": 20,
    "scores": [90, 85, 92],
    "is_active": True
}
# 값 접근
print(student["name"])          # 홍길동
print(student.get("age"))       # 20 (get: 키 없어도 오류 안 남)
print(student.get("grade", "없음"))  # 없음 (기본값 지정)
# 값 추가/수정
student["grade"] = "A"          # 새 키-값 추가
student["age"] = 21             # 기존 값 수정
print(student)
# 삭제
del student["is_active"]        # 키-값 삭제
removed = student.pop("grade")  # 삭제 + 반환
print(f"삭제된 값: {removed}")
# 딕셔너리 순회
print("\n키 목록:", list(student.keys()))
print("값 목록:", list(student.values()))
print("키-값 쌍:", list(student.items()))

# for문으로 순회
print()
for key, value in student.items():
    print(f"  {key}: {value}")
`````

## 💡 심화: 가변 객체(Mutable)와 불변 객체(Immutable)
- **가변 객체 (리스트, 딕셔너리):** 생성 후에도 내부 값을 변경할 수 있습니다. 변수에 할당할 때 값을 복사하지 않고 **메모리 주소**를 공유합니다.
- **불변 객체 (튜플, 문자열):** 한 번 생성되면 내부 값을 변경할 수 없습니다. 데이터의 무결성이 중요할 때(예: 좌표값, 설정값) 튜플을 사용합니다.

**[응용 실습] 자율주행 차량 센서 데이터 파싱(Dictionary & List)**
차량의 현재 상태가 아래와 같이 딕셔너리로 수집되었습니다.
1) 차량의 현재 속도를 추출하여 출력하세요.
2) 후방 레이더(rear)에 감지된 장애물 거리를 슬라이싱을 이용해 출력하세요.

`````python
# 실습용 데이터베이스 (API 응답 JSON 형태)
car_status = {
    "car_model": "Grandeur",
    "speed": 85.5,
    "radars": {
        "front": [15.2, 14.8, 14.5],
        "rear": [5.5, 4.2, 2.1]
    }
}

# 1) 차량 속도 출력
print(f"현재 속도: {car_status['speed']} km/h")

# 2) 후방 레이더의 가장 최근(마지막) 감지 거리 출력
print(f"후방 장애물 거리: {car_status['radars']['rear'][-1]} m")
`````

리스트는 데이터를 추가(`append`), 삽입(`insert`), 삭제(`remove`, `pop`)할 수 있는 유연한 자료형입니다. 딕셔너리와 함께 가장 많이 사용됩니다.

**[기초 강화 실습] 장바구니 리스트 관리**

# Python 함수(Functions)

이 노트북에서는 Python의 함수에 대해 학습합니다.

**함수(Function)** 란 특정 작업을 수행하는 **재사용 가능한 코드 블록**입니다.

**함수를 사용하는 이유:**
- **재사용성**: 한 번 작성하면 여러 곳에서 호출 가능
- **가독성**: 코드를 의미 있는 단위로 분리
- **유지보수**: 수정이 필요할 때 한 곳만 변경
- **모듈화**: 큰 프로그램을 작은 단위로 분리

## 1. 함수 정의와 호출

**기본 형태:**

```python
def 함수이름(매개변수1, 매개변수2, ...):
    """독스트링: 함수 설명 (선택사항)"""
    실행할 코드
    return 반환값  # 선택사항

# 함수 호출
결과 = 함수이름(인수1, 인수2, ...)

```

**용어 정리:**

* **매개변수(parameter)**: 함수 정의 시 선언하는 변수 이름 (입력구)
* **인수(argument)**: 함수 호출 시 전달하는 실제 값 (입력값)
* **반환값(return value)**: 함수가 호출자에게 돌려 주는값 (출력값)

**함수의 5가지 형태** (단계별 이해)

## 3. 매개변수 종류

Python 함수의 매개변수는 다양한 방식으로 지정할 수 있습니다.

### 3-1. 기본값 매개변수 (Default Parameter)

매개변수에 기본값을 설정하면, 호출 시 해당 인수를 생략할 수 있습니다.

> ⚠️ **규칙**: 기본값 매개변수는 반드시 **일반 매개변수 뒤**에 와야 합니다.

1.사용자로부터 정수 3개를 입력 받음  
2.입력 받은 변수의 합과 곱을 return하는 하나의 함수 작성  
3.입력받은 3개의 변수를 함수에 전달해서 return받은 결과 값을 화면에 표시  

````python
def calculate_sum_product(num1, num2, num3):
    total_sum = num1 + num2 + num3
    total_product = num1 * num2 * num3
    return total_sum, total_product

n1, n2, n3 = map(int, input("세 개의 정수를 입력하세요 (예: 10 20 30):").split())

sum_result, product_result = calculate_sum_product(n1, n2, n3)

print(f"정수들의 합은: {sum_result}")
print(f"정수들의 곱은: {product_result}")
````

````python
def greet(name, greeting="안녕하세요"):   # greeting에 기본값
    print(f"{greeting}, {name}님!")

greet("홍길동")                    # 기본값 사용 → 안녕하세요, 홍길동님!
greet("이순신", "반갑습니다")       # 기본값 대신 다른 값 사용

print()

# 실용 예시: 계산 함수
def calculate_bmi(weight, height, unit="metric"):
    """BMI 계산 함수
    weight: 몸무게 (kg)
    height: 키 (m)
    """
    bmi = weight / (height ** 2)
    if bmi < 18.5:
        status = "저체중"
    elif bmi < 23:
        status = "정상"
    elif bmi < 25:
        status = "과체중"
    else:
        status = "비만"
    return round(bmi, 1), status

bmi, status = calculate_bmi(70, 1.75)
print(f"BMI: {bmi} → {status}")
````

### 3-2. 키워드 인수 (Keyword Argument)

함수 호출 시 `매개변수명=값` 형식으로 명시적으로 인수를 전달할 수 있습니다.

````python
def introduce(name, age, city):
    print(f"이름: {name}, 나이: {age}, 도시: {city}")

# 위치 인수 (순서 중요)
introduce("홍길동", 25, "서울")

# 키워드 인수 (순서 무관)
introduce(age=25, city="서울", name="홍길동")  # 순서 바꿔도 OK

# 혼합 (위치 인수 먼저, 키워드 인수 나중)
introduce("홍길동", city="서울", age=25)
````

### 3-3. 가변 인수 `*args`와 `**kwargs`

| 형태 | 의미 | 타입 |
|------|------|------|
| `*args` | 위치 인수를 여러 개 받음 | 튜플 |
| `**kwargs` | 키워드 인수를 여러 개 받음 | 딕셔너리 |

````python
# *args: 임의 개수의 위치 인수
def sum_all(*args):
    """전달된 모든 숫자의 합을 반환"""
    print(f"args 타입: {type(args)}, 값: {args}")
    return sum(args)

print(sum_all(1, 2, 3))           # 6
print(sum_all(1, 2, 3, 4, 5))     # 15
print(sum_all(10, 20))            # 30

print()

# **kwargs: 임의 개수의 키워드 인수
def print_info(**kwargs):
    """전달된 키-값 정보를 출력"""
    print(f"kwargs 타입: {type(kwargs)}")
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

print_info(name="홍길동", age=25, city="서울")
print_info(product="노트북", price=150000, stock=5)

print()

# *args와 **kwargs 함께 사용
def mixed_func(a, b, *args, **kwargs):
    print(f"일반 인수: a={a}, b={b}")
    print(f"추가 위치 인수: {args}")
    print(f"키워드 인수: {kwargs}")

mixed_func(1, 2, 3, 4, 5, name="홍길동", score=95)
````

## 4. 변수 범위(Scope)

Python에서 변수는 **선언된 위치**에 따라 접근 가능한 범위가 결정됩니다.

| 범위 | 설명 |
|------|------|
| **지역 변수** (local) | 함수 내부에서 선언 → 함수 내에서만 접근 가능 |
| **전역 변수** (global) | 함수 외부에서 선언 → 프로그램 전체에서 접근 가능 |

````python
# 전역 변수
global_var = "나는 전역 변수"

def test_scope():
    local_var = "나는 지역 변수"   # 함수 내에서만 존재
    print(global_var)             # 전역 변수 읽기 가능
    print(local_var)

test_scope()
````

````python
# 전역 변수
global_var = "나는 전역 변수"

def test_scope():
    local_var = "나는 지역 변수"   # 함수 내에서만 존재
    print(global_var)             # 전역 변수 읽기 가능
    print(local_var)
    
test_scope()
print(local_var)
````

````python
# 함수 밖에서 지역 변수 접근 → 오류!
try:
    print(local_var)
except NameError as e:
    print(f"오류: {e}")

print()

# global 키워드: 함수 내에서 전역 변수 수정
count = 0

def increment():
    global count    # 전역 변수임을 선언
    count += 1

increment()
increment()
increment()
print(f"count = {count}")   # 3

print()

# 좋은 방법: global 대신 반환값 사용
def increment_v2(current_count):
    return current_count + 1

count2 = 0
count2 = increment_v2(count2)
count2 = increment_v2(count2)
print(f"count2 = {count2}")  # 2
````

## 5. 람다 함수(Lambda Function)

**이름 없는 함수**로, 간단한 함수를 한 줄로 표현할 수 있습니다.

```python
# 일반 함수
def add(x, y):
    return x + y

# 람다 함수 (동일한 기능)
add = lambda x, y: x + y
```

> 💡 람다 함수는 주로 `sorted()`, `map()`, `filter()` 등의 인수로 사용됩니다.

````python
# 기본 람다
add = lambda x, y: x + y
square = lambda x: x ** 2
is_even = lambda x: x % 2 == 0

print(add(3, 5))        # 8
print(square(4))        # 16
print(is_even(10))      # True

print()

# sorted()와 함께 사용 (정렬 기준 지정)
students = [
    {"name": "홍길동", "score": 85},
    {"name": "이순신", "score": 92},
    {"name": "김유신", "score": 78},
]

# 점수 기준으로 정렬
sorted_students = sorted(students, key=lambda s: s["score"], reverse=True)
for s in sorted_students:
    print(f"{s['name']}: {s['score']}점")

print()

# map()과 함께 사용 (변환)
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print("제곱:", squared)

# filter()와 함께 사용 (필터링)
evens = list(filter(lambda x: x % 2 == 0, range(1, 11)))
print("짝수:", evens)
````

## 6. 재귀 함수(Recursive Function)

함수가 **자기 자신을 호출**하는 방식입니다.  
반드시 **종료 조건(base case)** 이 있어야 합니다.

```
factorial(3)
  = 3 × factorial(2)
  = 3 × 2 × factorial(1)
  = 3 × 2 × 1          ← 종료 조건 (1! = 1)
  = 6
```

````python
# 팩토리얼 (n! = n × (n-1) × ... × 1)
def factorial(n):
    if n <= 1:          # 종료 조건 (base case)
        return 1
    return n * factorial(n - 1)  # 재귀 호출

print(f"5! = {factorial(5)}")   # 120
print(f"10! = {factorial(10)}") # 3628800

print()

# 피보나치 수열
def fibonacci(n):
    """n번째 피보나치 수 반환 (0, 1, 1, 2, 3, 5, 8, ...)"""
    if n <= 1:         # 종료 조건
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print("피보나치 수열 (처음 10개):", end=" ")
for i in range(10):
    print(fibonacci(i), end=" ")
print()
````

## 💡 심화: 가변 매개변수 (`*args` 와 `**kwargs`)
데이터의 개수가 정해져 있지 않을 때 사용합니다.
- `*args` (Arguments): 임의의 개수의 위치 인수를 **튜플(Tuple)** 로 받습니다.
- `**kwargs` (Keyword Arguments): 임의의 개수의 키워드 인수를 **딕셔너리(Dict)** 로 받습니다.

**[응용 실습] 데이터 분석 파이프라인 (가변 매개변수 활용)**
수집된 센서 데이터의 개수가 매번 다릅니다. 임의의 개수의 숫자를 입력받아(`*args`),
가장 큰 값, 가장 작은 값, 그리고 전체 평균을 계산하여 딕셔너리로 반환하는 함수 `analyze_sensor_data`를 작성하세요.

````python
# 정답 예시
def analyze_sensor_data(*args):
    if not args:
        return None
    
    max_val = max(args)
    min_val = min(args)
    avg_val = sum(args) / len(args)
    
    return {"최대": max_val, "최소": min_val, "평균": avg_val}

# 함수 호출 테스트
result1 = analyze_sensor_data(24.5, 25.1, 23.8, 26.2)
result2 = analyze_sensor_data(80, 95, 100)

print("센서 그룹 1 분석:", result1)
print("센서 그룹 2 분석:", result2)
````

## 💡 핵심 보완: `print`와 `return`의 차이점 이해

- `print()`: 단순히 화면에 값을 **보여주기만** 합니다. 다른 변수에 결과를 저장할 수 없습니다.
- `return`: 함수의 실행 결과를 **반환**하여, 다른 변수에 저장하거나 추가적인 계산에 사용할 수 있게 해줍니다.

**[기초 강화 실습] 부가세 계산기 함수**
물건의 원래 가격을 입력받아, 10%의 부가세가 포함된 최종 가격을 `return`하는 함수 `calculate_tax(price)`를 만드세요. 그 후 5000원짜리 물건의 최종 가격을 변수에 저장하여 출력해보세요.

````python
# 정답 예시
def calculate_tax(price):
    final_price = price * 1.1
    return int(final_price) # 계산된 값을 반환

# 함수 호출 및 결과 저장
item_price = 5000
result = calculate_tax(item_price)

print(f"원래 가격: {item_price}원")
print(f"부가세 포함 가격: {result}원")
````

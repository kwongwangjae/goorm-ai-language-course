```python
import matplotlib.pyplot as plt

plt.title("복수의 plot 명령을 한 그림에서 표현")

plt.plot(
    [1, 4, 9, 16],
    c="b",
    lw=5,
    ls="--",
    marker="o",
    ms=15,
    mec="g",
    mew=5,
    mfc="r"
)

plt.plot(
    [9, 16, 4, 1],
    c="k",
    lw=3,
    ls=":",
    marker="s",
    ms=10,
    mec="m",
    mew=5,
    mfc="c"
)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.linspace(-np.pi, np.pi, 256)

C, S = np.cos(X), np.sin(X)

plt.title("legend를 표시한 플롯")

plt.plot(X, C, ls="--", label="cosine")
plt.plot(X, S, ls=":", label="sine")

plt.legend(loc=0)
plt.show()
```

```python
np.random.seed(0)
f1 = plt.figure(figsize=(10,2)) # 캔버스 사이즈
plt.title("figure size : (10, 4)")
plt.plot(np.random.randn(100)) # 랜덤 그래픽
plt.show()
```

```python

x1 = np.linspace(0.0, 5.0)
x2 = np.linspace(0.0, 2.0)
y1 = np.cos(2 * np.pi * x1) * np.exp(-x1)
y2 = np.cos(2 * np.pi * x2)

ax1 = plt.subplot(2, 1, 1) ## 2행 1열 행렬의 첫번째(위쪽에 위치)
plt.plot(x1, y1, 'yo-')
plt.title('A tale of 2 subplots') 
plt.ylabel('Damped oscillation')
print (ax1)

ax2 = plt.subplot(2,1, 2) ## 2행 1열 행렬의 두번째(애래쪽에 위치)
plt.plot(x2, y2, 'r.-')
plt.xlabel('time (s) ')
plt.ylabel('Undamped' )
print (ax2)

plt.tight_layout()# 여백을 줄여중
plt.show()
```

```python
import time

# 현재 시간을 정수로 변환하여 시드값으로 설정
seed_value = int(time.time()) 
np.random.seed(seed_value)

plt.subplot(221) ## 2x2 행렬의 좌상단
plt.plot(np.random.rand(5))
plt.title("axes 1")

plt.subplot(222) ## 2x2 행렬의 우상단
plt.plot(np.random.rand(5))
plt.title("axes 2")

plt.subplot(223) ## 2x2 행렬의 좌하단
plt.plot(np.random.rand(5))
plt.title("axes 3")

plt.subplot(224) ## 2x2 행렬의 우하단
plt.plot(np.random.rand(5))
plt.title("axes 4")
```

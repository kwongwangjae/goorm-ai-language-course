# Python 조건문(Conditional Statements)

이 노트북에서는 Python의 조건문을 학습합니다.  
조건문은 특정 조건에 따라 **다른 코드 블록을 실행**할 수 있게 해주는 제어 구조입니다.

## 0. 조건문에 사용되는 연산자

### 비교 연산자
조건식에는 두 값을 비교하는 비교 연산자가 사용됩니다.  
비교 결과는 항상 `True` 또는 `False`입니다.

| 연산자 | 의미 | 예시 | 결과 |
|--------|------|------|------|
| `==` | 같다 | `5 == 5` | `True` |
| `!=` | 다르다 | `5 != 3` | `True` |
| `>` | 크다 | `5 > 3` | `True` |
| `<` | 작다 | `5 < 3` | `False` |
| `>=` | 크거나 같다 | `5 >= 5` | `True` |
| `<=` | 작거나 같다 | `3 <= 5` | `True` |

### 논리 연산자
여러 조건을 조합할 때 사용합니다.

| 연산자 | 의미 | 예시 | 결과 |
|--------|------|------|------|
| `and` | 둘 다 참이면 True | `5 > 3 and 2 < 4` | `True` |
| `or` | 하나라도 참이면 True | `5 > 10 or 2 < 4` | `True` |
| `not` | 참/거짓 반전 | `not True` | `False` |

### 멤버십 연산자
| 연산자 | 의미 | 예시 |
|--------|------|------|
| `in` | 포함되어 있으면 True | `"a" in "apple"` |
| `not in` | 포함되지 않으면 True | `5 not in [1,2,3]` |

````python
# 비교 연산자
print(5 == 5)   # True
print(5 != 3)   # True
print(5 > 3)    # True
print(5 < 3)    # False

print()
````

````python
# 논리 연산자
age = 20
print(age >= 18 and age < 65)   # True: 성인이면서 65세 미만
print(age < 10 or age >= 18)    # True: 어린이이거나 성인
print(not (age < 18))           # True: 미성년자가 아님

print()
````

````python
# 멤버십 연산자
fruits = ["apple", "banana", "cherry"]
print("apple" in fruits)        # True
print("grape" not in fruits)    # True

text = "Hello, Python!"
print("Python" in text)         # True
````

## 1. `if` 문

`if` 문은 조건이 **참(True)일 때만** 특정 코드를 실행합니다.

**기본 형태:**
```python
if 조건식:
    실행할 코드  # 반드시 들여쓰기(indent) 필요!
```

> ⚠️ **중요**: Python은 들여쓰기(indentation)로 코드 블록을 구분합니다.  
> 들여쓰기는 보통 **스페이스 4칸** 또는 **탭 1개**를 사용합니다.  
> 일반적으로(대부분의 코드에서는) **탭 1개** 를 사용합니다.  
> 들여쓰기가 틀리면 `IndentationError`가 발생합니다!  

````python
# 기본 if문
x = 10

if x > 5: #False
    print("x는 5보다 큽니다.")
    print(f"x의 값은 {x}입니다.")  # 같은 들여쓰기 = 같은 블록

print("이 줄은 항상 실행됩니다.")  # if문 밖

print()
````

````python

# 조건이 거짓인 경우 → 아무것도 실행 안 함
y = 3
if y > 10:
    print("이 줄은 출력되지 않습니다.")

print("y 조건 블록 끝")
````

## 2. `if-else` 문

`else`를 추가하면 조건이 **거짓(False)일 때** 실행할 코드를 지정할 수 있습니다.

```python
if 조건식:
    조건이 참일 때 실행
else:
    조건이 거짓일 때 실행
```

항상 둘 중 하나만 실행됩니다.

````python
# 기본 if-else
age = int(input("나이를 입력하세요: "))

if age >= 18:
    print("성인입니다. 입장 가능합니다.")
else:
    print("미성년자입니다. 입장이 제한됩니다.")
````

````python
# 수학 시험 합격/불합격
score = int(input("점수를 입력하세요: "))

if score >= 60:
    print(f"{score}점 - 합격!")
else:
    print(f"{score}점 - 불합격. {60 - score}점 부족합니다.")
````

````python
# 짝수/홀수 판별 (% 연산자 활용)
number = int(input("정수를 입력하세요: "))

if number % 2 == 0:
    print(f"{number}은(는) 짝수입니다.")
else:
    print(f"{number}은(는) 홀수입니다.")
````

## 3. `if-elif-else` 문

`elif`(else if)를 사용하면 **3가지 이상**의 경우를 처리할 수 있습니다.

```python
if 조건1:
    조건1이 참일 때
elif 조건2:
    조건1은 거짓, 조건2가 참일 때
elif 조건3:
    조건1,2는 거짓, 조건3이 참일 때
else:
    모든 조건이 거짓일 때
```

> 💡 **동작 원리**: 위에서부터 순서대로 조건을 확인하고, **처음으로 참이 되는 블록** 하나만 실행됩니다.  
> 나머지 조건들은 확인하지 않고 건너뜁니다.

````python
# 학점 계산 (점수에 따른 등급)
score = int(input("점수를 입력하세요 (0~100): "))

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"점수: {score}점, 학점: {grade}")
````

````python
# 계절 판별
month = int(input("월(1~12)을 입력하세요: "))

if month in [3, 4, 5]:
    season = "봄"
elif month in [6, 7, 8]:
    season = "여름"
elif month in [9, 10, 11]:
    season = "가을"
elif month in [12, 1, 2]:
    season = "겨울"
else:
    season = "잘못된 입력"

print(f"{month}월은 {season}입니다.")
````

## 4. 중첩 조건문(Nested If)

조건문 안에 또 다른 조건문을 중첩하여 사용할 수 있습니다.

```python
if 조건1:
    if 조건2:          # 조건1이 참일 때만 검사
        코드 블록
```

> ⚠️ **주의**: 중첩이 깊어질수록 코드 가독성이 떨어집니다.  
> 가능하면 `and` 연산자로 조건을 합쳐서 중첩을 줄이는 것이 좋습니다.

````python
# 중첩 조건문 예시: 로그인 검증
username = input("아이디를 입력하세요: ")
password = input("비밀번호를 입력하세요: ")

correct_user = "admin"
correct_pass = "1234"

if username == correct_user:
    if password == correct_pass:
        print("로그인 성공! 환영합니다.")
    else:
        print("비밀번호가 틀렸습니다.")
else:
    print("존재하지 않는 아이디입니다.")

print()

# 중첩 대신 and 조건으로 간소화
if username == correct_user and password == correct_pass:
    print("로그인 성공! (and 방식)")
else:
    print("로그인 실패! (and 방식)")
````

## 5. 삼항 연산자 (조건부 표현식)

간단한 `if-else`를 **한 줄**로 작성하는 방법입니다.

```python
결과 = 참일 때 값  if  조건식  else  거짓일 때 값
```

````python
# 일반 if-else
x = 10
if x > 0:
    sign = "양수"
else:
    sign = "음수 또는 0"
    
print(sign)

# 삼항 연산자로 한 줄에 표현
sign = "양수" if x > 0 else "음수 또는 0"
print(sign)

print()

# 활용 예시
score = 85
result = "합격" if score >= 60 else "불합격"
print(f"점수 {score}점: {result}")

# 짝수/홀수
n = int(input("숫자 입력: "))
parity = "짝수" if n % 2 == 0 else "홀수"
print(f"{n}은(는) {parity}입니다.")
````

## 💡 심화: 단락 평가 (Short-circuit Evaluation)
논리 연산자(`and`, `or`)를 사용할 때, 파이썬은 결과가 확정되면 나머지 조건은 평가하지 않습니다.
- `A and B`: A가 False면 B는 확인하지 않고 False 반환. (가장 무거운 연산을 뒤로 배치하는 것이 유리)
- `A or B`: A가 True면 B는 확인하지 않고 True 반환.

**[응용 실습] 스마트 홈 IoT 센서 제어**
온도(temperature)와 습도(humidity) 센서 값을 입력받아 다음 조건에 따라 장비를 가동하세요.
- 온도 30도 이상이면서 습도 70% 이상: "에어컨 및 제습기 가동"
- 온도 30도 이상이면서 습도 70% 미만: "에어컨만 가동"
- 그 외의 경우: "현재 상태 유지"

````python
# 정답 예시
temp = float(input("현재 온도: "))
humidity = float(input("현재 습도(%): "))

if temp >= 30 and humidity >= 70:
    print("경보: 에어컨 및 제습기 가동")
elif temp >= 30:
    print("상태: 에어컨만 가동")
else:
    print("상태: 현재 상태 유지")
````

## 💡 핵심 보완: 조건문 작성 팁 (`=` vs `==`, 그리고 들여쓰기)
1. **할당과 비교 혼동:** 변수에 값을 넣을 때는 `=`, 값이 같은지 비교할 때는 `==`를 사용합니다.
2. **들여쓰기(Indentation):** 파이썬은 들여쓰기로 실행될 코드 블록을 구분합니다. 조건문 아래의 코드는 반드시 띄어쓰기 4칸(또는 Tab)을 맞춰야 합니다.

**[기초 강화 실습] 놀이공원 요금 계산기**

- 12세 이하: 15,000원
- 13세 ~ 18세: 20,000원
- 19세 이상: 30,000원

````python
# 정답 예시
age = int(input("나이를 입력하세요: "))

if age <= 12:
    print("어린이 요금: 15,000원 입니다.")
elif age <= 18:
    print("청소년 요금: 20,000원 입니다.")
else:
    print("성인 요금: 30,000원 입니다.")
````

**[문제 4]** 사용자로부터 **상품 구매 금액과 회원 등급**을 입력받아 최종 결제 금액을 계산하세요.

회원 등급 할인:
- "VIP" → 20% 할인
- "GOLD" → 10% 할인
- "SILVER" → 5% 할인
- 그 외 → 할인 없음

단, **구매 금액이 100000원 이상이면 추가로 5% 할인**을 적용하세요.

최종 결제 금액을 출력하세요.

````python
# 구매 금액 입력
amount = int(input("구매 금액을 입력하세요: "))
# 회원 등급 입력
grade = input("회원 등급을 입력하세요 (VIP/GOLD/SILVER): ")
# 기본 할인율 설정
discount_rate = 0
check = True
# 회원 등급에 따른 할인
if grade == "VIP":
    discount_rate = 0.20
elif grade == "GOLD":
    discount_rate = 0.10
elif grade == "SILVER":
    discount_rate = 0.05
else:
    print("잘못된 입력, 할인 0")
    check = False
# 추가 할인 조건
if check == True and amount >= 100000:
    discount_rate += 0.05
# 최종 결제 금액 계산
final_price = amount * (1 - discount_rate)
# 결과 출력
print("최종 결제 금액:", int(final_price), "원")
````

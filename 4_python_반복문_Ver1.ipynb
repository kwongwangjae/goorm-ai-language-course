# Python 반복문(Loops)

이 노트북에서는 Python의 반복문을 학습합니다.  
반복문은 **동일한 코드를 여러 번 실행**할 때 사용하는 제어 구조입니다.

| 종류 | 사용 시기 |
|------|----------|
| `while` | 조건이 참인 동안 반복 (반복 횟수 모를 때) |
| `for` | 시퀀스를 순회하거나 횟수가 정해진 반복 |

---
## 1. `while` 문

**형태:**
```python
while 조건식:
    반복할 코드
    조건 변경 코드  # 무한루프 방지!
```

> ⚠️ **주의**: 조건을 항상 참으로 만들면 **무한 루프**에 빠집니다.  
> 반드시 루프 내에서 조건이 거짓이 될 수 있도록 값을 변경해야 합니다.

````python
# 기본 while문
x = 0
while x < 5:
    print(f"x = {x}")
    x += 1   # 이 줄이 없으면 무한루프!

print(f"루프 종료, 최종 x = {x}")
````

````python
# 4 3 2 1 0 코드 각자 작성
x = 4
while x > -1:
    print(f"x = {x}")
    x -= 1   # 이 줄이 없으면 무한루프!

print(f"루프 종료, 최종 x = {x}")
````

````python
# while 1~10 합 복합 연산
````

````python
# 활용: 합계 계산 (1~10까지 합)
total = 0
n = 1
while n <= 10:
    total += n  #total = total + n
    n += 1
print(f"1~100 합계: {total}")
````

````python
# 활용: 사용자 입력 검증 (올바른 입력이 들어올 때까지 반복)
while True:
    age = input("나이를 입력하세요 (1~120): ")
    if age.isdigit() and 1 <= int(age) <= 120:
        age = int(age)
        print(f"입력된 나이: {age}")
        break
    else:
        print("잘못된 입력입니다. 다시 입력하세요.")
````

## 2. `for` 문

**형태:**
```python
for 변수 in 시퀀스:
    반복할 코드
```

시퀀스(리스트, 튜플, 문자열, range 등)의 각 요소를 순서대로 꺼내어 변수에 할당합니다.

````python
my_lst = [3,1,7,9]
for i in my_lst:
    print(i)

print("종료")
````

### 2-1. `range()` 함수

숫자 시퀀스를 생성하는 함수입니다.

| 형태 | 의미 | 예시 |
|------|------|------|
| `range(n)` | 0 ~ n-1 | `range(5)` → 0,1,2,3,4 |
| `range(start, end)` | start ~ end-1 | `range(2, 6)` → 2,3,4,5 |
| `range(start, end, step)` | step 간격 | `range(0, 10, 2)` → 0,2,4,6,8 |

````python
# range 기본
for i in range(5):
    print(i, end=" ")   # 0 1 2 3 4
print()
````

````python
# range(start, end)
for i in range(1, 6):
    print(i, end=" ")   # 1 2 3 4 5
print()

# range(start, end, step) - 2씩 증가
for i in range(0, 11, 2):
    print(i, end=" ")   # 0 2 4 6 8 10
print()

# range - 역방향 (감소)
for i in range(5, 0, -1):
    print(i, end=" ")   # 5 4 3 2 1
print()

print()

````

````python
#사용자에게 구구단의 "단"을 입력 받아서 해당 구구단 화면에 표시
# for문으로 구구단
dan = int(input("단을 입력하세요 :"))
print(f"--- {dan}단 ---")
for i in range(1, 10):
    print(f"{dan} × {i} = {dan * i}")
````

````python
# for문으로 구구단
dan = 3
print(f"--- {dan}단 ---")
for i in range(1, 10):
    print(f"{dan} × {i} = {dan * i}")
````

### 2-2. 시퀀스 순회

````python
# 리스트 순회
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"과일: {fruit}")
````

````python
# 문자열 순회 (한 글자씩)
for char in "Python":
    print(char, end="-")
````

````python
# enumerate() - 인덱스와 값을 함께
for idx, fruit in enumerate(fruits):
    print(f"{idx}번: {fruit}")

# enumerate 시작 번호 지정
for idx, fruit in enumerate(fruits, start=1):
    print(f"{idx}번째: {fruit}")
````

````python
# zip() - 두 리스트를 병렬로 순회
names = ["홍길동", "이순신", "김유신"]
scores = [85, 92, 78]

for name, score in zip(names, scores):
    print(f"{name}: {score}점")

print()

# 딕셔너리 순회
student = {"국어": 90, "영어": 85, "수학": 92}
for subject, score in student.items():
    print(f"{subject}: {score}점")
print(f"평균: {sum(student.values()) / len(student):.1f}점")
````

## 3. `break`와 `continue`

| 키워드 | 동작 |
|--------|------|
| `break` | 반복문을 **즉시 종료** |
| `continue` | 현재 반복을 **건너뛰고** 다음 반복으로 |
| `else` | 반복문이 **정상 종료**됐을 때 실행 (break 없이) |

````python
# break 예시: 특정 조건에서 반복 중단
print("--- break ---")
for i in range(10):
    if i == 5:
        print("..5를 만나 중단!")
        break
    print(i, end=" ")
````

````python
# 활용: 리스트에서 특정 값 찾기
numbers = [4, 7, 2, 9, 1, 5, 8]
target = 9
for idx, num in enumerate(numbers):
    if num == target:
        print(f"{target}을(를) {idx}번 위치에서 발견!")
        break
````

````python
# continue 예시: 특정 조건 건너뛰기
print("--- continue (홀수만 출력) ---")
for i in range(1, 11):
    if i % 2 == 0:   # 짝수면 건너뜀
        continue
    print(i, end=" ")
print()

print()

# for-else: break 없이 정상 종료 시 else 실행
print("--- for-else ---")
search = [1, 2, 3, 4, 5]
target = 10

for num in search:
    if num == target:
        print(f"{target} 발견!")
        break
else:
    print(f"{target}을(를) 찾지 못했습니다.")  # break 없이 루프 종료 시 실행
````

## 4. 중첩 루프(Nested Loop)

````python
# 기본 중첩 루프
for i in range(3):
    for j in range(3):
        print(f"({i},{j})", end=" ")
    print()  # 안쪽 루프 끝나면 줄바꿈
````

2X1=2

....
9x9=81

````python
for i in range(2,10):  #2~9
    for j in range(1,10):   #1~9
        #print(f"{i} x {j} = {i * j}", end="\n")
        print(f"{i} x {j} = {i * j}")
    print()
````

````python


# 활용: 구구단 전체 출력
for dan in range(2, 10):
    print(f"--- {dan}단 ---")
    for i in range(1, 10):
        print(f"{dan}×{i}={dan*i:2d}", end="  ")
    print()

print()

````

````python
# 활용: 별 패턴 출력
n = 5
for i in range(1, n + 1):
    print("★" * i)

print()

# 역삼각형
for i in range(n, 0, -1):
    print("★" * i)
````

## 5. 리스트 컴프리헨션 (List Comprehension)

for문을 **한 줄**로 간결하게 작성하여 리스트를 생성하는 방법입니다.

```python
# 일반 for문
result = []
for 변수 in 시퀀스:
    result.append(표현식)

# 리스트 컴프리헨션 (동일한 결과)
result = [표현식 for 변수 in 시퀀스]

# 조건 추가
result = [표현식 for 변수 in 시퀀스 if 조건]
```

````python
# 1~10
#result = [x                  for x in range(1, 11)]
result = [x for x in range(1, 11)]
print("1~10 :", result)
````

````python
# 1~10 제곱 리스트
squares = [x**2 for x in range(1, 11)]
print("제곱:", squares)
````

````python
# 1~20 중 짝수만
'''
for x in range(1, 21) 
    if x % 2 == 0
'''
#evens = [x*2 for x in range(1, 21) if x % 2 == 0]
evens = [x for x in range(1, 21) if x % 2 == 0]
print("짝수:", evens)
````

````python
# 문자열 처리
fruits = ["apple", "banana", "cherry", "avocado"]
a_fruits = [f.upper() for f in fruits if f.startswith("a")]
print("a로 시작하는 과일(대문자):", a_fruits)

# 온도 변환 (섭씨 → 화씨)
celsius = [0, 10, 20, 30, 40]
fahrenheit = [c * 9/5 + 32 for c in celsius]
print("섭씨:", celsius)
print("화씨:", fahrenheit)
````

## 💡 심화: 파이썬만의 특별한 반복문 (`for-else` / `while-else`)
루프가 중간에 `break`로 끊기지 않고 **정상적으로 끝까지 수행되었을 때만** `else` 블록이 실행됩니다.
특정 데이터를 찾거나, 재시도 로직을 구현할 때 매우 유용합니다.

**[응용 실습] 하드웨어 장비 연결 재시도 로직 (`while` & `break`)**
장비와 연결을 시도합니다. 최대 5회까지 재시도하며, 사용자가 'y'를 입력하면 "연결 성공"을 출력하고 루프를 탈출합니다.
5회를 모두 시도했는데도 연결되지 않으면 `else`문을 활용해 "장비 점검 요망"을 출력하세요.

````python
# 정답 예시
attempt = 1
max_try = 5

while attempt <= max_try:
    response = input(f"연결 시도 {attempt}/{max_try} - 연결되었습니까? (y/n): ")
    if response.lower() == 'y':
        print("하드웨어 연결 성공!")
        break
    attempt += 1
else:
    print("5회 연결 실패. 장비 점검 요망 (Error Connection)")
````

## 💡 핵심 보완: 흐름 제어문 (`break`와 `continue`)
- `break`: 반복문을 즉시 **완전히 종료**하고 빠져나옵니다. (주로 무한루프 종료 시 사용)
- `continue`: 이번 반복의 남은 코드를 건너뛰고, **다음 반복으로 바로 넘어갑니다.**

**[기초 강화 실습] 안전한 비밀번호 입력기**
비밀번호가 'python123'입니다. 사용자가 올바른 비밀번호를 입력할 때까지 계속 질문하는 루프를 만드세요. 맞추면 "로그인 성공"을 출력하고 루프를 탈출(`break`)합니다.

````python
# 정답 예시
correct_pw = "python123"

while True:
    pw = input("비밀번호를 입력하세요: ")
    if pw == correct_pw:
        print("로그인 성공! 환영합니다.")
        break
    else:
        print("비밀번호가 틀렸습니다. 다시 시도하세요.\n")
````
